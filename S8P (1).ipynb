```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create x array
x = np.linspace(0, 1, 100)

# Define functions for sine and cosine
def get_sin(val):
    return np.sin(val)

def get_cos(val):
    return np.cos(val)
```

```python
# Create a figure with 1 row and 2 columns
# figsize=(10, 4) makes it wide enough for two side-by-side plots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Left Panel: sin(x)
axes[0].plot(x, get_sin(x))
axes[0].set_title("sin(x)")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")

# Right Panel: cos(x)
axes[1].plot(x, get_cos(x))
axes[1].set_title("cos(x)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("g(x)")

# Adjust layout so labels don't overlap
plt.tight_layout()

# Save the figure as a PDF
plt.savefig("trig_plots.pdf")

# Display
plt.show()
```
