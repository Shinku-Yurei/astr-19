```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate 1000 random numbers uniformly distributed between 0 and 1
random_numbers = np.random.rand(1000)
```

```python
# Create the histogram with 100 bins
plt.hist(random_numbers, bins=100, edgecolor='black', alpha=0.7)

# Label the axes
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution of 1000 Random Numbers")

# Save the figure as a PDF
plt.savefig("random_histogram.pdf")

# Display the plot
plt.show()
```
