```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate 1000 random numbers from an exponential distribution
# scale=1.0 is the mean (average interval)
data = np.random.exponential(scale=1.0, size=1000)
```

```python
# Create the histogram with 100 bins
plt.hist(data, bins=100, edgecolor='black', color='skyblue')

# Label the axes
plt.xlabel("Interval Value")
plt.ylabel("Frequency")
plt.title("Exponential Distribution (1000 Samples)")

# Save the figure as a PDF
plt.savefig("exponential_histogram.pdf")

# Display the plot
plt.show()
```
